##### Question 7: ATM Menu
Create a simple ATM system using a while loop.

The initial balance is:

50000

Display this menu repeatedly:

1. Check Balance
2. Deposit
3. Withdraw
4. Exit

The program should continue until the user selects 4.

Implement the following:
- Check balance
- Deposit money
- Withdraw money
- Prevent withdrawal if the amount is greater than the balance
- Exit the program 

```python
balance = 50000
while True:
    print("1: Check Balance")
    print("2: Deposit")
    print("3: Withdraw")
    print("4: Exit")
    select = int(input("Tell your response:- "))

    if select == 1:
        print(f"Current Balance:- RS {balance}")
    elif select == 2:
        deposit = int(input("Enter deposit amount:- "))
        balance += deposit
        print("Amount deposit successfully.")
        print(f"Update balance is:- RS {balance}")
    elif select == 3:
        withdraw = int(input("Enter withdraw amount:- "))
        if withdraw <= balance:
            balance -= withdraw
            print("Amount withdraw successfully.")
            print(f"Update balance is:- RS {balance}")
        else:
            print("Insufficient amount.")
    elif select == 4:
        break
    else:
        print("Invalid Response")
```

##### Question 8: Number Guessing Game
Create a number guessing game.

Set the secret number to:

secret_number = 25

Ask the user to guess the number.

Use a while loop.

If the guess is:
- Greater than 25 → "Too High"
- Less than 25 → "Too Low"
- Equal to 25 → "Correct!"

Also count how many attempts the user made.

Example:

Guess: 30

Too High

Guess: 20

Too Low

Guess: 25

Correct!

Attempts: 3

```python
secretNumber = 25
tries = 0
while True:
    guess = int(input("Guess the number:- "))
    tries += 1
    if guess == secretNumber:
        print("Correct!\n"
              f"Attempts:- {tries}")
        break
    elif guess > secretNumber:
        print("Too High.")
    elif guess < secretNumber:
        print("Too Low.")
    else:
        print("Very Close.")
```

##### Question 9: Student Marks Entry System
A teacher wants to enter student marks continuously.

Use a while loop.

The program should keep asking for marks until the teacher enters:

-1

For every student's marks:

• 50 or above → Pass

• Below 50 → Fail

At the end, display:

Total Students

Total Passed

Total Failed

The -1 should not be counted as a student's marks.

```python
totalStudents = 0
totalPassed = 0
totalFailed = 0
while True:
    marks = int(input("Enter student marks (-1 to stop):- "))
    if marks == -1:
        break
    elif marks >= 50:
        totalStudents += 1
        totalPassed += 1
        print("Pass")
    elif marks < 50:
        totalStudents += 1
        totalFailed += 1
        print("Fail")

print(f"Total Students:- {totalStudents}\n"
      f"Total Passed:- {totalPassed}\n"
      f"Total Failed:- {totalFailed}")
```

##### Question 11: Student Performance Management System
A teacher wants to create a system for analyzing the performance of multiple students.

First ask:

Enter number of students:

For each student, ask:
- Student name
- Marks in Python
- Marks in Mathematics
- Marks in English

Calculate:

Total Marks

Percentage

Result Rules

A student is considered Pass only if:
- Overall percentage is at least 50%
- AND marks in every subject are at least 40.

Otherwise, the student is Fail.

Grade Rules

For students who pass:

90+ → A+

80-89 → A

70-79 → B

60-69 → C

50-59 → D

While processing students, calculate:

- Number of passed students
- Number of failed students
- Highest percentage
- Lowest percentage
- Class total marks
- Class average percentage

Final Display:

===== CLASS REPORT =====

Total Students: 10

Passed: 7

Failed: 3

Highest Percentage: 94%

Lowest Percentage: 42%

Class Average: 76.5%

Important requirement

Use a loop to process all students.

Do not manually write the same code for every student

```python
totalStudents = int(input("Enter total students:- "))
passed = 0
failed = 0
classTotalMarks = 0
totalPercentage = 0

for i in range(totalStudents):
    print(f"Students {i + 1}")
    name = input("Enter student name:- ")
    pythonMarks = int(input("Enter your python marks:- "))
    mathMarks = int(input("Enter your math marks:- "))
    englishMarks = int(input("Enter your english marks:- "))

    totalMarks = pythonMarks + mathMarks + englishMarks
    percentage = (totalMarks / 300) * 100

    if percentage >= 50 and pythonMarks >= 40 and mathMarks >= 40 and englishMarks >= 40:
        print("Result:- Pass")
        passed += 1
        if percentage >= 90:
            print("Grade:- A+")
        elif percentage >= 80:
            print("Grade:- A")
        elif percentage >= 70:
            print("Grade:- B")
        elif percentage >= 60:
            print("Grade:- C")
        else:
            print("Grade:- D")
    else:
        failed += 1
        print("Result:- Fail")

    classTotalMarks += totalMarks
    totalPercentage += percentage

    if i == 0:
        highestPercentage = percentage
        lowestPercentage = percentage
    else:
        if percentage > highestPercentage:
            highestPercentage = percentage
        if percentage < lowestPercentage:
            lowestPercentage = percentage

if totalStudents > 0:
    classAverage = totalPercentage / totalStudents

    print("===== CLASS REPORT =====\n"
          f"Total Students:- {totalStudents}\n"
          f"Passed Students:- {passed}\n"
          f"Failed Students:- {failed}\n"
          f"Class Total Marks:- {classTotalMarks}\n"
          f"Highest Percentage:- {highestPercentage} %\n"
          f"Lowest Percentage:- {lowestPercentage} %\n"
          f"Class Average Percentage:- {classAverage} %")
else:
    print("No students entered.")
```

##### Question 12: Restaurant Ordering & Billing System
You are developing a restaurant ordering system.

The restaurant has the following menu:

MENU
1. Burger Rs. 500
2. Pizza Rs. 1000
3. Biryani Rs. 350
4. Fries Rs. 250
5. Drink Rs. 150
6. Checkout

The program should repeatedly display the menu and allow the customer to order multiple items.

The program should continue until the customer selects:
6. Checkout

When an item is selected

Add its price to the total bill.

Also keep track of how many items the customer ordered.

For example:

Burger added

Pizza added

Drink added

Invalid Input

If the customer enters something other than 1-6:

Invalid choice. Please try again.

The program should not terminate.

Discount Rules

When the customer checks out:

Bill >= 5000

→ 20% discount

Bill >= 3000

→ 15% discount

Bill >= 1500

→ 10% discount

Below 1500

→ No discount

Delivery Charges

After discount:

Final bill >= 3000

→ Free Delivery

Final bill < 3000

→ Rs. 200 Delivery Charges


Finally display:

===== FINAL BILL =====

Total Items: 5

Original Bill: Rs. 3500

Discount: Rs. 525

Delivery Charges: Rs. 0

Final Bill: Rs. 2975

Thank you for ordering!

Important requirement

Use a while loop because the number of orders is unknown. The customer should be able to

keep ordering until they choose checkout.

```python
totalBill = 0
totalItems = 0
while True:
    print("===== RESTURANT MENU =====\n"
          "1: Burger - Rs 500\n"
          "2: Pizza - Rs 1000\n"
          "3: Biryani - Rs 350\n"
          "4: Fries - Rs 250\n"
          "5: Drink - Rs 150\n"
          "6: Checkout")

    choice = int(input("Tell your response:- "))
    if choice == 1:
        totalBill += 500
        totalItems += 1
        print("Burger added.")
    elif choice == 2:
        totalBill += 1000
        totalItems += 1
        print("Pizza added.")
    elif choice == 3:
        totalBill += 350
        totalItems += 1
        print("Biryani added.")
    elif choice == 4:
        totalBill += 250
        totalItems += 1
        print("Fries added.")
    elif choice == 5:
        totalBill += 150
        totalItems += 1
        print("Drink added.")
    elif choice == 6:
        break
    else:
        print("Invalid response. Please try again!")

if totalBill >= 5000:
    discount = (totalBill * 20) / 100
elif totalBill >= 3000:
    discount = (totalBill * 15) / 100
elif totalBill >= 1500:
    discount = (totalBill * 10) / 100
else:
    discount = 0

discountBill = totalBill - discount

if discountBill >= 3000:
    deliveryCharges = 0
else:
    deliveryCharges = 200

finalBill = discountBill + deliveryCharges

print("===== FINAL BILL =====\n"
      f"Total Items:- {totalItems}\n"
      f"Original Bill:- Rs {totalBill}\n"
      f"Discount:- {discount}\n"
      f"Delivery Charges:- {deliveryCharges}\n"
      f"Final Bill:- Rs {finalBill}\n"
      "Thankyou for ordering!")
```
